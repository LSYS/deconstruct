```python
import os
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', 50)
# dummy switch to drop variables after being processed
DROP = True
```

```python
dfVOTER = (
    pd.read_csv(os.path.join('intermediate', 'voter.csv'))
    .drop(['zip','inputstate_2016','izip_2016','inputzip_2018'], axis=1)
    .dropna(subset=["fips"])
)
dfVOTER.head()
```

### Sexism measures
* Measures used:
    * sexism1
    * sexism2
    * sexism3 [reversed]
    * sexism4
    * sexism5 
    * sexism6 [reversed]

* Individual sexism: Range 1-4 (most sexist to least)
* Total sexism: Range 1-24 (least to most sexist)
* sexism change: Range -23 to 23 (decrease to increase in sexism)

**Output:**
* sexism16 (range 1-24, least to most sexist)
* sexism18 (range 1-24, least to most sexist) 
* sexism_change (range -23 to 23, increase in sexism)

```python
# convert sexism3 and sexism6 to go from 1-4 (most sexist to not)
dfVOTER['sexism3_2016_reversed'] = 5 - dfVOTER['sexism3_2016']
dfVOTER['sexism6_2016_reversed'] = 5 - dfVOTER['sexism6_2016']

# convert sexism3 and sexism6 to go from 1-4 (most sexist to not)
dfVOTER['sexism3_2018_reversed'] = 5 - dfVOTER['sexism3_2018']
dfVOTER['sexism6_2018_reversed'] = 5 - dfVOTER['sexism6_2018']

# Create aggregate for 2016
dfVOTER['sexism16'] = (dfVOTER['sexism1_2016'] +  dfVOTER['sexism2_2016'] + dfVOTER['sexism3_2016_reversed'] 
                       +  dfVOTER['sexism4_2016'] +  dfVOTER['sexism5_2016'] +  dfVOTER['sexism6_2016_reversed'])
dfVOTER['sexism16'] = 25 - dfVOTER['sexism16']

# Create aggregate for 2018
dfVOTER['sexism18'] = (dfVOTER['sexism1_2018'] +  dfVOTER['sexism2_2018'] + dfVOTER['sexism3_2018_reversed'] 
                       +  dfVOTER['sexism4_2018'] +  dfVOTER['sexism5_2018'] +  dfVOTER['sexism6_2018_reversed'])
dfVOTER['sexism18'] = 25 - dfVOTER['sexism18']

# Change in sexism
dfVOTER['sexism_change'] = dfVOTER['sexism18'] - dfVOTER['sexism16']

# Drop original sexism variables
if DROP:
    dfVOTER.drop(['sexism1_2016', 'sexism2_2016','sexism3_2016','sexism4_2016','sexism5_2016','sexism6_2016', 
                  'sexism3_2016_reversed', 'sexism6_2016_reversed',
                  'sexism1_2018', 'sexism2_2018','sexism3_2018','sexism4_2018','sexism5_2018','sexism6_2018', 
                  'sexism3_2018_reversed', 'sexism6_2018_reversed'], axis=1, inplace=True)
```

### Ideology
**Output:**
* ideo_baseline_18 (Range -4 to 4,  increase in conservatism)
* ideo_16_18 (Range -4 to 4, increase in conservatism)

#### Ideo5
Measures used:
    * ideo5_baseline 
    * ideo5_2016
    * ideo5_2018
        Scale:
            1. Very liberal
            2. Liberal 
            3. MOderate
            4. Conservative
            5. Very Conservative

```python
# Convert ideo5 measures to exlude those with values 6 (not sure)
dfVOTER['ideo5_baseline_processed'] = np.where(dfVOTER['ideo5_baseline']<6, dfVOTER['ideo5_baseline'], np.nan)
dfVOTER['ideo5_2016_processed'] = np.where(dfVOTER['ideo5_2016']<6, dfVOTER['ideo5_2016'], np.nan)
dfVOTER['ideo5_2018_processed'] = np.where(dfVOTER['ideo5_2018']<6, dfVOTER['ideo5_2018'], np.nan)

dfVOTER['ideo_baseline_18'] = dfVOTER['ideo5_2018_processed'] - dfVOTER['ideo5_baseline_processed']
dfVOTER['ideo_16_18'] = dfVOTER['ideo5_2018_processed'] - dfVOTER['ideo5_2016_processed']
```

#### pid3
Measures used:
    * pid3_baseline
    * pid3_2016
    * pid3_2018
        Scale:
            1. Rep
            2. Ind
            3. Dem
** Output:**
* pid3_baseline_processed2 (1 Dem 2 Others/Ind 3 Rep)
* pid3_2016_processed2 (1 Dem 2 Others/Ind 3 Rep) 
* pid3_2018_processed2 (1 Dem 2 Others/Ind 3 Rep)
* pid_baseline_18 (-2 to 2, increasing in Dem alignment)
* pid_16_18 (-2 to 2, increasing in Dem alignment)

** baseline **

```python
# Merge 3 (Ind) and 4 (Other)
dfVOTER['pid3_baseline_processed1'] = np.where(((dfVOTER['pid3_baseline']==3) | (dfVOTER['pid3_baseline']==4)), 
                                               3, dfVOTER['pid3_baseline'])

# Drop == 5 (not sure)
dfVOTER['pid3_baseline_processed2'] = np.where(dfVOTER['pid3_baseline_processed1']<4, dfVOTER['pid3_baseline_processed1'], np.nan)

# Change Rep to 3, Ind/Others to 2, so that the scale goes from 1 Dem 2 Ind/others 3 Rep
# to capture a "political indifference"
dfVOTER['pid3_baseline_rep_holder'] = np.where(dfVOTER['pid3_baseline_processed2']==2, 1, 0)
dfVOTER['pid3_baseline_ind_holder'] = np.where(dfVOTER['pid3_baseline_processed2']==3, 1, 0)

dfVOTER['pid3_baseline_processed2'] = np.where(dfVOTER['pid3_baseline_rep_holder']==1, 3, dfVOTER['pid3_baseline_processed2'])
dfVOTER['pid3_baseline_processed2'] = np.where(dfVOTER['pid3_baseline_ind_holder']==1, 2, dfVOTER['pid3_baseline_processed2'])
```

** 2016 **

```python
# Merge 3 (Ind) and 4 (Other)
dfVOTER['pid3_2016_processed1'] = np.where(((dfVOTER['pid3_2016']==3) | (dfVOTER['pid3_2016']==4)), 
                                               3, dfVOTER['pid3_2016'])

# Drop == 5 (not sure)
dfVOTER['pid3_2016_processed2'] = np.where(dfVOTER['pid3_2016_processed1']<4, dfVOTER['pid3_2016_processed1'], np.nan)

# Change Rep to 3, Ind/Others to 2, so that the scale goes from 1 Dem 2 Ind/others 3 Rep
# to capture a "political indifference"
dfVOTER['pid3_2016_rep_holder'] = np.where(dfVOTER['pid3_2016_processed2']==2, 1, 0)
dfVOTER['pid3_2016_ind_holder'] = np.where(dfVOTER['pid3_2016_processed2']==3, 1, 0)

dfVOTER['pid3_2016_processed2'] = np.where(dfVOTER['pid3_2016_rep_holder']==1, 3, dfVOTER['pid3_2016_processed2'])
dfVOTER['pid3_2016_processed2'] = np.where(dfVOTER['pid3_2016_ind_holder']==1, 2, dfVOTER['pid3_2016_processed2'])
```

** 2018 **

```python
# Merge 3 (Ind) and 4 (Other)
dfVOTER['pid3_2018_processed1'] = np.where(((dfVOTER['pid3_2018']==3) | (dfVOTER['pid3_2018']==4)), 
                                               3, dfVOTER['pid3_2018'])

# Drop == 5 (not sure)
dfVOTER['pid3_2018_processed2'] = np.where(dfVOTER['pid3_2018_processed1']<4, dfVOTER['pid3_2018_processed1'], np.nan)

# Change Rep to 3, Ind/Others to 2, so that the scale goes from 1 Dem 2 Ind/others 3 Rep
# to capture a "political indifference"
dfVOTER['pid3_2018_rep_holder'] = np.where(dfVOTER['pid3_2018_processed2']==2, 1, 0)
dfVOTER['pid3_2018_ind_holder'] = np.where(dfVOTER['pid3_2018_processed2']==3, 1, 0)

dfVOTER['pid3_2018_processed2'] = np.where(dfVOTER['pid3_2018_rep_holder']==1, 3, dfVOTER['pid3_2018_processed2'])
dfVOTER['pid3_2018_processed2'] = np.where(dfVOTER['pid3_2018_ind_holder']==1, 2, dfVOTER['pid3_2018_processed2'])
```

```python
# Create change variables
dfVOTER['pid_baseline_18'] = dfVOTER['pid3_2018_processed2'] - dfVOTER['pid3_baseline_processed2']
dfVOTER['pid_16_18'] = dfVOTER['pid3_2018_processed2'] - dfVOTER['pid3_2016_processed2']
```

```python
if DROP:
    dfVOTER.drop(['ideo5_baseline', 'ideo5_2016', 'ideo5_2018', 
                  'pid3_baseline_processed1',
                  'pid3_2016_processed1', 
                  'pid3_2018_processed1', 
                  'pid3_baseline_rep_holder', 'pid3_baseline_ind_holder',
                  'pid3_2016_rep_holder', 'pid3_2016_ind_holder',
                  'pid3_2018_rep_holder', 'pid3_2018_ind_holder',
                  'pid3_baseline', 'pid3_2016', 'pid3_2018'], axis=1, inplace=True)
```

### Vote
measures used:
* vote18_2018 (If an election for U.S. Congress were being held today, who would you vote for in your district)
* presvote16post_2016 (Who did you vote for in the election for President?)
* vote18_other_2018 (If you had to choose, would you
lean more toward Rep or Dem)

**Output**:
* vote_16 (1 = Rep, 2 = Dem)
* vote_18 (1 = Rep, 2 = Dem)
* vote2_18 [uses additional "forcing" question] (1 = Rep, 2 = Dem)
* vote_change  (range -1 to 1, -1 if change from Rep to Dem, 1 if change from Dem to Rep)
* vote_change2 [uses additional "forcing" question] (range -1 to 1, -1 if change from Rep to Dem, 1 if change from Dem to Rep)

#### 2016

```python
# Impute nan (other) 4 (not sure) 5 (would not vote)
dfVOTER['vote_16'] = np.where(dfVOTER['presvote16post_2016']<3, dfVOTER['presvote16post_2016'], np.nan)

# Change 1 from CLinton to 2 (Dem), 2 from Trump to 1 (Rep)
dfVOTER['clinton_placeholder'] = np.where(dfVOTER['vote_16']==1, 1, 0)
dfVOTER['trump_placeholder'] = np.where(dfVOTER['vote_16']==2, 1, 0)

dfVOTER['vote_16'] = np.where(dfVOTER['clinton_placeholder']==1, 2, dfVOTER['vote_16'])
dfVOTER['vote_16'] = np.where(dfVOTER['trump_placeholder']==1, 1, dfVOTER['vote_16'])
```

#### 2018

```python
# Impute nan (other) 4 (not sure) 5 (would not vote)
dfVOTER['vote_18'] = np.where(dfVOTER['vote18_2018']<3, dfVOTER['vote18_2018'], np.nan)

# Impute not sure/would not vote using vote18_other_2018
dfVOTER['vote2_18'] = np.where(dfVOTER['vote18_other_2018']==1, 1, dfVOTER['vote_18'])

dfVOTER['vote2_18'] = np.where(dfVOTER['vote18_other_2018']==2, 2, dfVOTER['vote_18'])
```

#### Change

```python
dfVOTER['vote_change'] = dfVOTER['vote_18'] - dfVOTER['vote_16']
dfVOTER['vote_change2'] = dfVOTER['vote2_18'] - dfVOTER['vote_16']
```

```python
if DROP:
    dfVOTER.drop(['presvote16post_2016', 'vote18_2018', 
                  'clinton_placeholder', 'trump_placeholder'], 
                 axis=1, inplace=True)
```
