```python
fig, ax = plt.subplots(figsize=(6, 4), constrained_layout=True)

sns.histplot(
    df["bot_score"],
    bins=50,
    stat="density",
    color="0.5",
    edgecolor=".3",
    alpha=.9,
    linewidth=0.7,
    ax=ax,
)
ax.tick_params(axis="y", which="both", left=False, labelleft=False)
ax.set_xlabel("Botometer X score for probability of bot-like behavior")
plt.savefig("out/botometer_score_distribution.pdf", bbox_inches="tight")
```

```python
for threshold in [.1,.2,.3,.4,0.43, 0.5, 0.6, 0.7, 0.76, 0.8, 0.9]:
    df = df.dropna(subset=["bot_score"])

    n_above = len(df.query(f"bot_score > {threshold}"))
    n_below = len(df.query(f"bot_score <= {threshold}"))
    n_total = n_above + n_below

    pct_above = (n_above / n_total) * 100 if n_total else 0
    pct_below = (n_below / n_total) * 100 if n_total else 0

    print(f"Threshold: {threshold:.2f}")
    print(f"  Above threshold: {n_above:,} ({pct_above:.1f}%)")
    print(f"  Below or equal: {n_below:,} ({pct_below:.1f}%)")
    print("-" * 40)
```
