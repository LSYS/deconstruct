```python
import os
import pandas as pd
import re
from tqdm import tqdm
```

### Preparing tweets

```python
tweetsFilesList = ['jan2018.csv',
                   'feb2018.csv',
                   'mar2018.csv',                   
                   'apr2018.csv',
                   'may2018.csv', 
                   'jun2018.csv',                   
                   'jul2018.csv',
                   'aug2018.csv',
                   'sep2018.csv',
                   'oct2018.csv',
                   'nov2018.csv']

# get all tweets in 2018 (Jan 2018 to 6 Nov 2018)
dfTweets = pd.DataFrame()

for fileName in tweetsFilesList:
    
    _df = pd.read_csv(os.path.join('tweets-data', fileName), 
                      index_col=0)
    dfTweets = dfTweets.append(_df, ignore_index=True)
```

### Getting top hashtags

```python
hashPattern = "#\s{0,1}[a-zA-Z0-9]+"

hashtagDict = dict()

for ix in tqdm(range(0, len(dfTweets))):
    tweet = dfTweets.at[ix, 'content']
    hashtagList = re.findall(hashPattern, tweet)
    
    for hashtag in hashtagList:
        # remove hashtag and lowercase
        hashtag = hashtag.replace('#','').lower().strip()
        
        # add count to dictionary
        if hashtag not in hashtagDict:
            hashtagDict[hashtag] = 1
        else:
            hashtagDict[hashtag] += 1
```

```python
dfHashtags = pd.DataFrame.from_dict(hashtagDict, orient='index')
dfHashtags.reset_index(inplace=True)
dfHashtags.columns = ['hashtag', 'count']
```

```python
dfHashtags.sort_values('count', ascending=False, inplace=True)
```

```python
dfHashtags.to_csv('hashtags.csv', index=False)
```
