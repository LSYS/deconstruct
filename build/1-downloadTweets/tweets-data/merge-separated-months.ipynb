```python
import pandas as pd
import os.path
```

## Merge 3 separate sep2018 tweets data

```python
folderpath = 'misc'
```

```python
dfSep1 = pd.read_csv(os.path.join(folderpath, 'sep2018-01to10.csv'), index_col=0)

dfSep2 = pd.read_csv(os.path.join(folderpath, 'sep2018-10to20.csv'), index_col=0)

dfSep3 = pd.read_csv(os.path.join(folderpath, 'sep2018-20to30.csv'), index_col=0)

print(len(dfSep1) + len(dfSep2) + len(dfSep3))
```

```python
dfSep = dfSep1.append([dfSep2, dfSep3], ignore_index=True)
len(dfSep)
```

```python
dfSep.drop_duplicates(dfSep.columns, inplace=True, keep='first')
len(dfSep)
```

```python
dfSep.to_csv('sep2018.csv')
```
